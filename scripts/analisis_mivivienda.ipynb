```python
ruta_1 = r"C:\Users\Kenji\Desktop\Fondo Mivivienda\mivivienda.csv"
vivienda = pd.read_csv(ruta_1)
display(vivienda.head())
```

```python
# Numero de filas, columnas
print(vivienda.shape)
```

```python
# Para analizar columnas, axis = 1
nr_nulo = vivienda.isna().all(axis=1).sum()
print(nr_nulo)
# dropna() elimina todos los registros que tengan nulos
if nr_nulo > 0:
    vivienda = vivienda.dropna()
    print(vivienda.shape)
```

```python
# Verificar si el registro tiene datos duplicados
nr_duplicado = vivienda.duplicated().sum()
print(nr_duplicado)
# Eliminamos filas duplicadas
if nr_duplicado > 0:
    vivienda = vivienda.drop_duplicates()
    # Mostramos filas actuales (0)
    print(vivienda.shape[0])
print(vivienda.shape)
```

```python
# Eliminamos la columna FECHA_CORTE
vivienda = vivienda.drop("FECHA_CORTE", axis=1)
print("Filas válidas finales: ", vivienda.shape[0])
```
